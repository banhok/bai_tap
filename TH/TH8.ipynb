```python
#W7A16
from collections import Counter

def mang_dep(a):
    tso = Counter(a)
    xoa = 0
    print(tso)
    print(tso.items())
    for x, tso_x in tso.items():
        if x == tso_x:
            continue
        elif tso_x > x:
            xoa += tso_x - x
        else:
            xoa += tso_x

    return "YES" if xoa == 0 else f"NO {xoa}"


a = list(map(int, input().split()))
print(mang_dep(a))
```

```python
#W7A17
def phan_tu(a):

    ptu = None
    dem = 0

    for x in a:
        if dem == 0:
            ptu = x
            dem = 1
        elif x == ptu:
            dem += 1
        else:
            dem -= 1
        
    if ptu is not None and a.count(ptu) > len(a) // 2:
        return ptu
    return None

a = list(map(int, input().split()))
print(phan_tu(a))


```

```python
#W7A19
def merge_sorted_arrays(A, B):
    result = []
    i, j = 0, 0
    
    
    ascending = A[0] <= A[-1]
    
    while i < len(A) and j < len(B):
        if ascending:
            if A[i] <= B[j]:
                result.append(A[i])
                i += 1
            else:
                result.append(B[j])
                j += 1
        else:  
            if A[i] >= B[j]:
                result.append(A[i])
                i += 1
            else:
                result.append(B[j])
                j += 1
    
   
    result.extend(A[i:])
    result.extend(B[j:])
    
    return result


```

```python
#W7A22
from collections import Counter

def most_common_name(names):

    last_names = [name.split()[-1] for name in names]
    freq = Counter(last_names)
    max_count = max(freq.values())
    candidates = [name for name, count in freq.items() if count == max_count]    
    
    return min(candidates)

N=int(input())
names = [input().strip() for _ in range(N)]

print(most_common_name(names))
```

```python
#W7A23
def count_names(s):
    for ch in ",.!?:;":
        s = s.replace(ch," ")
    words = s.split()
    return sum(1 for w in words if len(w) >= 2 and w.isalpha() and w.istitle())

s = input()
count_names(s)
```

```python
#W7A24
def remove_adjacent_duplicates(s):
    words = s.split()
    res = [words[0]] if words else []
    for w in words[1:]:
        if w != res[-1]:
            res.append(w)
    return ' '.join(res)

s = input()
x=remove_adjacent_duplicates(s)
print(x)
```

```python
#W7A25
def max_k_digits(s, k):
    stack = []
    drop = len(str(s)) - k
    for ch in str(s):
        while drop and stack and stack[-1] < ch:
            stack.pop()
            drop -= 1
        stack.append(ch)
    return ''.join(stack[:k])

s, k = map(int, input().split())
x=max_k_digits(s, k)
print(x)
```

```python
#W7A26
import re

def sum_and_avg(s):
    nums = list(map(int, re.findall(r'\d+', s)))
    total = sum(nums)
    avg = total // len(nums) if nums else 0
    return total, avg

s = input()
x=sum_and_avg(s)
print(*x)
```

```python
#W7A27
def ok(s: str) -> bool:
    ngoac_mo = []
    axa = {')': '(', '}': '{', ']': '['}

    for ngoac in s:
        if ngoac in axa:
            ngoac_mo_tu = ngoac_mo.pop() if ngoac_mo else '#'
            if axa[ngoac] != ngoac_mo_tu:
                return False
        else:
            ngoac_mo.append(ngoac)

    return not ngoac_mo



if __name__ == "__main__":
    test_strings = ["()", "()[]{}", "(]", "([)]", "{[]}", "((()))", "({[()]})", "((())", "())"]
    for s in test_strings:
        print(f"{s}: {ok(s)}")

```

```python
#W7A28
import math
def is_composite(n):
    if n <= 3:
        return False 
    for i in range(2, math.isqrt(n) + 1):
        if n % i == 0:
            return True
    return False


def longest_composite_sequence(N):
    max_len = 0
    max_start = -1

    current_len = 0
    current_start = -1

    for i in range(1, N + 1):
        if is_composite(i):
            if current_len == 0:
                current_start = i
            current_len += 1
        else:
            if current_len > max_len:
                max_len = current_len
                max_start = current_start
            current_len = 0

    if current_len > max_len:
        max_len = current_len
        max_start = current_start

    return max_start, max_len


if __name__ == '_main_':
    N = int(input())
    start, length = longest_composite_sequence(N)
    if length != 0 :
        for i in range(start, start + length):
            print(i, end=" ")
```

```python
#W7A29
def LIC(arr):
    res, temp = [], []
    for i in range(len(arr)):
        if i == 0 or arr[i] > arr[i-1]:
            temp.append(arr[i])
        else:
            if len(temp) > len(res):
                res = temp[:]
            temp = [arr[i]]
    if len(temp) > len(res):
        res = temp
    return res

arr=list(map(int, input().split()))
print(*LIC(arr))

```

```python
#W7A30
def LIS(a):
    n=len(a)
    dp=[1]*n
    pre=[-1]*n
    for i in range(n):
        for j in range(i):
            if a[j]<a[i] and dp[j]+1>dp[i]:
                dp[i]=dp[j]+1
                pre[i]=j
    k=dp.index(max(dp))
    d=[]
    while k!=-1:
        d.append(a[k])
        k=pre[k]
    return d[::-1]

a=list(map(int,input().split()))
print(*LIS(a))
```

```python
#W7A31
def doi_cho_min(s):
    s = list(str(s))
    n = len(s)
    nho_nhat = int("".join(s))
    kq = s[:]

    for i in range(n):
        for j in range(i+1, n):
            tmp = s[:]
            tmp[i], tmp[j] = tmp[j], tmp[i]
            if tmp[0] != '0':
                so_moi = int("".join(tmp))
                if so_moi < nho_nhat:
                    nho_nhat = so_moi
                    kq = tmp[:]
    return int("".join(kq))

x = int(input())
print(doi_cho_min(x))
```
