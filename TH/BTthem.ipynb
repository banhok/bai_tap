```python
#cho dãy arr và số k, tìm dãy có độ dài lớn nhất có trung bình cộng lớn hơn k

def longest_subarray_avg_gt_k(arr, k):
    n = len(arr)
    max_len = 0
    best_l = best_r = -1

    for i in range(n):
        s = 0
        for j in range(i, n):
            s += arr[j]
            length = j - i + 1
            if s / length > k:
                if length > max_len:
                    max_len = length
                    best_l = i
                    best_r = j

    if max_len == 0:
        return [] 
    return arr[best_l:best_r+1]

arr = list(map(int, input().split()))
k = int(input())
longest_subarray_avg_gt_k(arr, k)
```

```python
##cho dãy arr và số k, tìm dãy có độ dài nhỏ nhất có trung bình cộng lớn hơn k

def shortest_subarray_avg_gt_k(arr, k):
    n = len(arr)
    min_len = float('inf')
    best_l = best_r = -1

    for i in range(n):
        s = 0
        for j in range(i, n):
            s += arr[j]
            length = j - i + 1
            if s / length > k:
                if length < min_len:
                    min_len = length
                    best_l = i
                    best_r = j
                break 
    if best_l == -1:
        return []
    return arr[best_l:best_r+1]

arr = list(map(int, input().split()))
k = int(input())
shortest_subarray_avg_gt_k(arr, k)
```

```python
#to hop chap k cua n bang de quy

def ckn(n, k):
    if k > n:
        return -1
    if k == n or k == 0:
        return 1
    else:
        return ckn(n-1, k-1) + ckn(n-1, k)
    

ckn(15, 3)
```

```python
#fibonaci

def fibonacci(n):
    if n == 0: return 0
    if n == 1: return 1
    return fibonacci(n-1) + fibonacci(n-2)

print(fibonacci(6))

```

```python
# collatz

def collatz(x):
    print(x, end=" ")  
    if x == 1:         
        return
    if x % 2 == 0:    
        collatz(x // 2)
    else:             
        collatz(3 * x + 1)

collatz(7)
```

```python
# bai toan thap ha noi
def tower(N, a, b, c):
    if N == 1:
        print(f"{a} -> {c}")
        return
    
    tower(N - 1, a, c, b)

    tower(1, a, b, c)
    
    tower(N - 1, b, a, c)

if __name__ == "__main__":
    N = int(input("Nhập số đĩa: "))
    a, b, c = 1, 2, 3
    tower(N, a, b, c)
```
