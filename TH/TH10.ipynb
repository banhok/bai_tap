```python
#W9A7 
def canPlaceFlowers(flowerbed, k):
    n = len(flowerbed)
    for i in range(n):
        if flowerbed[i] == 0:
            left = (i == 0) or (flowerbed[i-1] == 0)
            right = (i == n-1) or (flowerbed[i+1] == 0)
            if left and right:
                flowerbed[i] = 1
                k -= 1
                if k == 0:
                    return True
    return k <= 0

flowerbed = list(map(int, input().split()))
k = int(input())
print(canPlaceFlowers(flowerbed, k))

```

```python
#W9A8
def is_path_crossing(moves: str) -> bool:
    x, y = 0, 0
    visited = {(0, 0)}
    for move in moves:
        if move == 'R':
            x += 1
        elif move == 'L':
            x -= 1
        elif move == 'U':
            y += 1
        elif move == 'D':
            y -= 1
        if (x, y) in visited:
            return True
        visited.add((x, y))
    return False

moves = input()
print(is_path_crossing(moves))
```

```python
#W9A9 
def Median(nums):
    a, b, c, d, e = nums
    if a > b: a, b = b, a
    if c > d: c, d = d, c
    if a > c: a, c = c, a
    if b > d: b, d = d, b
    if b > c: b, c = c, b
    if c > e: c, e = e, c
    return c
```

```python
#W9A10
import bisect

def findRadius(houses, heaters):
    houses.sort()
    heaters.sort()
    res = 0
    for h in houses:
        
        idx = bisect.bisect_left(heaters, h)
        
        left = h - heaters[idx-1] if idx > 0 else float('inf')
        
        right = heaters[idx] - h if idx < len(heaters) else float('inf')

        nearest = min(left, right)
        res = max(res, nearest)
    return res


```
