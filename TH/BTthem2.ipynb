```python
#bai1 (leo cầu thang)
def ways(n):
    if n == 1: return 1
    if n == 2: return 2
    a, b = 1, 2
    for _ in range(3, n+1):
        a, b = b, a+b
    return b

n = int(input())
print(ways(n))
```

```python
#bai2 (dãy con liên tiếp dài nhất có tbc > k)
def ok(a, k):
    n = len(a)
    best = []

    for i in range(n):
        s = 0
        for j in range(i, n):
            s += a[j]
            if s / (j - i + 1) > k:
                if j - i + 1 > len(best):
                    best = a[i:j+1]

    if best:
        return best
    return "khong co day hop le"


a = list(map(int, input().split()))
k = int(input())
print(ok(a, k))
```

```python
#bai3 (phân tích tsnt)
import math
def pt(n):
    kq=""
    for i in range(2, math.isqrt(n)+1):
        if n%i==0:
            mu=0
            while n%i==0:
                mu+=1
                n//=i
            kq+=f"{i}^{mu}"
            if n!=1:
                kq+='*'
    if n>1:
        kq += f"{n}^1"
    return kq


n = int(input())
print(pt(n))
```

```python
#bai4
def clean(text):
    not_ok = ".,!?;:"
    ok = ""
    for kt in text:
        if kt not in not_ok:
            ok += kt
    return ok

def dem(cau, tu):
    words = clean(cau).split()
    print(words)
    target = clean(tu).split()
    print(target)
    dem = 0


    for i in range(len(words) - len(target) + 1):
        if words[i:i+len(target)] == target:
            dem += 1
    return dem

cau = input()
tu = input()
print(dem(cau, tu))

    
```
