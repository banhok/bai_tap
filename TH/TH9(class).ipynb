```python
#W8A1
class Cylinder:
    def __init__(self, r, h):
        self.r, self.h = r, h
        self.pi = 3.14
    


    def Sbm(self):
        return 2*self.pi*self.r*(self.r + self.h)
    
    def V(self):
        return self.pi*self.r**2 * self.h
    
r, h = map(float, input().split())
htru = Cylinder(r, h)

print(f"{htru.Sbm():.2f} {htru.V():.2f}")
```

```python
#W8A2
class Date:
    def __init__(self, d, m , y):
        self.d, self.m, self.y = d, m , y

    def nam_nh(self):
        return (self.y % 400 == 0) or (self.y %4 == 0 and self.y % 100 != 0)
    
    def hop_le(self):
        d_in_m = [31, 29 if self.nam_nh() else 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]

        if self.m <1 or self.m > 12:
            return False
        if self.d < 1 or self.d > d_in_m[self.m - 1]:
            return False
        return True
    
    def ngay_tiep(self):
        d_in_m = [31, 29 if self.nam_nh() else 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]

        self.d += 1
        if self.d > d_in_m[self.m - 1]:
            self.d = 1
            self.m+=1
            if self.m > 12:
                self.m = 1
                self.y+=1
        
        return Date(self.d, self.m, self.y)
    

date_str = input().strip()
d, m, y = map(int, date_str.split("/"))
date = Date(d, m , y)
if not date.hop_le():
    print("INVALID")
else:
    next_date = date.ngay_tiep()
    print(f"{next_date.d:02d}/{next_date.m:02d}/{next_date.y:02d}")    
        
```

```python
#W8A3
import math
class Point2():
    def __init__(self, x, y):
        self.x, self.y = x, y

    def gtd(self):
        return self.x == 0 and self.y == 0
    
    def Oy(self):
        return self.x == 0 and self.y != 0
    
    def Ox(self):
        return self.x != 0 and self.y == 0
    
    def dist_O(self):
        return math.sqrt(self.x**2 + self.y**2)
    
    def __str__(self):
        if self.gtd(): 
            pos = "M là gốc tọa độ" 
        elif self.Oy(): 
            pos = "M thuộc trục tung" 
        elif self.Ox(): 
            pos = "M thuộc trục hoành" 
        else: 
            pos = "M không thuộc Ox và Oy" 
        return f"Điểm M({self.x}, {self.y}), {pos}, khoảng cách đến O: {self.dist_O():.2f}"

    

x, y = map(int, input().split())
M = Point2(x, y)
print(M)


```

```python
#W8A5
class ShoppingCart():

    def __init__(self):#1
        self.gio_hang = []

    def them_sp(self, ten, gia):#2
        self.gio_hang.append({"ten": ten, "gia": gia})

    def xoa_sp(self, ten):#3
        for sp in self.gio_hang:
            if sp["ten"] == ten:
                self.gio_hang.remove(sp)
                return True
        return False
    
    def gio_hang_rong(self):#4
        return len(self.gio_hang) == 0
    
    def tong_tien(self):#5
        return sum(sp["gia"] for sp in self.gio_hang)
    
    def hien_thi(self):#6
        if self.gio_hang_rong():
            print("Gio hang rong")
        else:
            for sp in self.gio_hang:
                print(f"{sp['ten']}: {sp['gia']}")

    def xoa_het(self):#7
        self.gio_hang.clear()

#Menu
cart = ShoppingCart()

while True:
    print("\nMenu:")
    print("2. Thêm sản phẩm")
    print("3. Xóa sản phẩm")
    print("4. Kiểm tra giỏ hàng rỗng")
    print("5. Tính tổng tiền giỏ hàng")
    print("6. Hiển thị toàn bộ sản phẩm")
    print("7. Xóa toàn bộ giỏ hàng")

    choice = input("Nhập lựa chọn (2-7): ")

    if choice == "2":
        ten = input("Nhập tên sản phẩm: ")
        gia = float(input("Nhập giá sản phẩm: "))
        cart.them_sp(ten, gia)
    elif choice == "3":
        ten = input("Nhập tên sản phẩm cần xóa: ")
        if cart.xoa_sp(ten):
            print("Đã xóa sản phẩm.")
        else:
            print("Không tìm thấy sản phẩm.")
    elif choice == "4":
        if cart.gio_hang_rong():
            print("Giỏ hàng rỗng.")
        else:
            print("Giỏ hàng không rỗng.")
    elif choice == "5":
        print(f"Tổng tiền giỏ hàng: {cart.tong_tien()}")
    elif choice == "6":
        cart.hien_thi()
    elif choice == "7":
        cart.xoa_het()
        print("Đã xóa toàn bộ giỏ hàng.")
    else:
        print("Lựa chọn không hợp lệ.")

    exit_choice = input("Có muốn thoát khỏi chương trình không? (yes/no): ")
    if exit_choice.lower() == "yes":
        print("Chương trình kết thúc.")
        break
```

```python
#W8A6
class Rectangle():
    def __init__(self,w,h):
        self.w = w
        self.h = h
    def scale(self,k):
        self.w *= k
        self.h *= k
    def perimeter(self):
        return (self.w + self.h) * 2
    def area(self):
        return self.w * self.h
w,h = map(float,input().split())
n = Rectangle(w,h)
k = int(input("phóng to k lần : "))
n.scale(k)
print(f"{n.area():.2f} {n.perimeter():.2f}")
```

```python
#W8A7
import math
class Fraction():
    def __init__(self,tu,mau):
        self.a = tu 
        self.b = mau
    def checkphanso(self):
        return isinstance(self.a, int) and isinstance(self.b, int) and self.b != 0
    def rutgonphanso(self,m,n):
        x = math.gcd(m, n)
        m //= x
        n //= x
        if n < 0:
            m, n = -m, -n
        return Fraction(m,n)
    def __add__(self,other):
        if not  other.checkphanso() or not self.checkphanso():
            return "phân số  ko hợp lệ"
        m,n = 0,0
        if self.b == other.b:
            m = self.a + other.a
            n = self.b
        else:
            m = self.a*other.b + other.a*self.b
            n = self.b * other.b
        return self.rutgonphanso(m,n)
    def __sub__(self,other):
        if not  other.checkphanso() or not self.checkphanso():
            return "phân số  ko hợp lệ"
        m,n = 0,0
        if self.b == other.b:
            m = self.a - other.a
            n = self.b
        else:
            m = self.a*other.b - other.a*self.b
            n = self.b * other.b
        return self.rutgonphanso(m,n)
    def __mul__(self,other):
        m,n = 0,0
        if not  other.checkphanso() or not self.checkphanso():
            return "phân số  ko hợp lệ"
        else:
            m = self.a * other.a
            n = self.b * other.b
        return self.rutgonphanso(m,n)
    def __truediv__(self,other):
        m,n = 0,0
        if not  other.checkphanso() or not self.checkphanso() or other.a == 0:
            return "phân số  ko hợp lệ"
        else:
            m = self.a * other.b
            n = self.b * other.a
        return self.rutgonphanso(m,n)
    def __str__(self):
        return f"{self.a} / {self.b}"
a,b = map(int,input().split())
s1 = Fraction(a,b)
i,j = map(int,input().split())
s2 = Fraction(i,j)
print("1 : cộng")
print("2 : trừ")
print("3 : nhân")
print("4 : chia")
chon = int(input())
if chon == 1:
    print(s1+s2)
if chon == 2:
    print(s1-s2)
if chon == 3:
    print(s1*s2)
if chon == 4:
    print(s1/s2)
```

```python
#W8A8
class BankAccount():
    def __init__(self,owner,balance):
        self.ten = owner
        self.tien = balance
    def deposit(self,x):
        self.tien += x
    def withdraw(self,x):
        self.tien -= x
    def __str__(self):
        return f"{self.tien:.2f}"
a = input("nhập tên người dùng : ")
b = float(input("nhập số tiền trong tài khoản: "))
n = BankAccount(a,b)
k = int(input("Nhập số theo tác : "))
for i in range(k):
    choice,x = input("nhập giao dịch: ").split()
    x = float(x)
    if choice == "DEPOSIT":
        n.deposit(x)
    else:
        if n.tien >= x:
            n.withdraw(x)
print(n)
```

```python
#W8A10
class library():
    def __init__(self):
        self.giasach =[]
    def themsach(self,title,author,year):
        a = [title,author,year]
        self.giasach.append(a)
    def count_by_author(self,name):
        cnt = 0
        for i in range(len(self.giasach)):
            if self.giasach[i][1] == name:
                cnt += 1
        return cnt
    def find_by_year(self,y):
        cnt = 0
        for i in range(len(self.giasach)):
            if self.giasach[i][2] == y:
                cnt += 1
        return cnt
nam = library()
k = int(input())
for x in range(k):
    m,n = input().split(" ",1)
    if m == "ADD":
        title,author,year = n.split(";")
        nam.themsach(title,author,int(year))
    elif m == "COUNT":
        print(nam.count_by_author(n))
    elif m == "COUNTYEAR":
        print(nam.find_by_year(int(n)))
```

```python
#W8A11
import math
class Angle:
    def __init__(self, A, B, C, D):
        self.A = A
        self.B = B
        self.C = C
        self.D = D
    def vector(self, P, Q):
        return [Q[0]-P[0], Q[1]-P[1], Q[2]-P[2]]
    def cross(self, a, b):
        return [
            a[1]*b[2] - a[2]*b[1],
            a[2]*b[0] - a[0]*b[2],
            a[0]*b[1] - a[1]*b[0]
        ]
    def dot(self, a, b):
        return a[0]*b[0] + a[1]*b[1] + a[2]*b[2]
    def length(self, v):
        return math.sqrt(self.dot(v, v))
    def angle_between_planes(self):
        AB = self.vector(self.A, self.B)
        AC = self.vector(self.A, self.C)
        BC = self.vector(self.B, self.C)
        BD = self.vector(self.B, self.D)
        n1 = self.cross(AB, AC)
        n2 = self.cross(BC, BD)
        cos_phi = abs(self.dot(n1, n2)) / (self.length(n1) * self.length(n2))
        phi = math.degrees(math.acos(cos_phi))
        return phi
A = list(map(float, input().split()))
B = list(map(float, input().split()))
C = list(map(float, input().split()))
D = list(map(float, input().split()))
goc = Angle(A, B, C, D)
print(f"{goc.angle_between_planes():.2f}")
```

```python
#W8A12
class ComplexNumber():
    def __init__(self,doubleinitRealPart,doubleinitImaginaryPart):
        self.a = doubleinitRealPart
        self.b = doubleinitImaginaryPart
    def ComplexNumbermultiply(self,other):
        thuc = self.a * other.a - self.b*other.b
        ao = self.a * other.b + self.b * other.a
        return ComplexNumber(thuc,ao)
    def __str__(self):
        return f"{self.a} {self.b}"
a,b = map(int,input("nhập lần lượt phần thực và phần ảo cảu số phức : ").split())
c,d = map(int,input("nhập lần lượt phần thực và phần ảo cảu số phức : ").split())
s1 = ComplexNumber(a,b)
s2 = ComplexNumber(c,d)
print(s1.ComplexNumbermultiply(s2))
```

```python
#W8A13
import math
class Triangle():
    def __init__(self,canh1,canh2,canh3):
        self.a = canh1
        self.b = canh2
        self.c = canh3
    def doublegetArea(self):
        a, b, c = self.a, self.b, self.c
        if a <= 0 or b <= 0 or c <= 0 or a + b <= c or a + c <= b or b + c <= a:
            return -1
        p = (a + b + c) / 2
        return math.sqrt(p * (p - a) * (p - b) * (p - c))
canh1,canh2,canh3 = map(int,input("Nhập ll 3 cạnh của tam giác: ").split())
n = Triangle(canh1,canh2,canh3)
if n.doublegetArea() == -1:
    print("invalid")
else:
    print(f"{n.doublegetArea():.2f}")
        
```
