```python
#W9A1 
def merge_sorted_arrays(A, B):
    result = []
    i, j = 0, 0
    
    
    ascending = A[0] <= A[-1]
    
    while i < len(A) and j < len(B):
        if ascending:
            if A[i] <= B[j]:
                result.append(A[i])
                i += 1
            else:
                result.append(B[j])
                j += 1
        else:  
            if A[i] >= B[j]:
                result.append(A[i])
                i += 1
            else:
                result.append(B[j])
                j += 1
    
    result.extend(A[i:])
    result.extend(B[j:])
    
    return result
```

```python
#W9A2
def kthLargest(matrix, k):
    n, m = len(matrix), len(matrix[0])
    
    def countLessEqual(x):
        
        count, row, col = 0, n - 1, 0
        while row >= 0 and col < m:
            if matrix[row][col] <= x:
                count += row + 1
                col += 1
            else:
                row -= 1
        return count
    
    low, high = matrix[0][0], matrix[-1][-1]
    while low < high:
        mid = (low + high) // 2
        if countLessEqual(mid) < k:
            low = mid + 1
        else:
            high = mid
    return low
```

```python
#W9A3.1
def smallest_swap(n: int) -> int:
    s = list(str(n))
    for i in range(len(s)):
        
        j = min(range(i+1, len(s)), default=-1, key=lambda k: s[k])
        if j != -1 and s[j] < s[i]:
            s[i], s[j] = s[j], s[i]
            return int("".join(s))
    
    s[-1], s[-2] = s[-2], s[-1]
    return int("".join(s))

n = int(input())
print(smallest_swap(n))
```

```python
#W9A3.2
def largest_swap(n: int) -> int:
    s = list(str(n))
    for i in range(len(s)):
        j = max(range(i+1, len(s)), default=-1, key=lambda k: s[k])
        if j != -1 and s[j] > s[i]:
            s[i], s[j] = s[j], s[i]
            return int("".join(s))
    s[-1], s[-2] = s[-2], s[-1]
    return int("".join(s))
```

```python
#W9A4
def can_transform(a: str, b: str) -> str:
    j = 0
    for ch in a:
        if j < len(b) and ch.upper() == b[j]:
            if ch.isupper() or ch.upper() == b[j]:
                j += 1
        elif ch.isupper():
            return "NO"
    return "YES" if j == len(b) else "NO"


Q = int(input().strip())
for _ in range(Q):
    a = input().strip()
    b = input().strip()
    print(can_transform(a, b))
```

```python
#W9A5 
def is_subarray(list1, list2):
    n, m = len(list1), len(list2)
    if n > m:
        return False
    for i in range(m - n + 1):
        if list2[i:i+n] == list1:
            return True
    return False

```

```python
#W9A6 
def next_permutation(arr):
    i = len(arr) - 2
    while i >= 0 and arr[i] >= arr[i+1]: i -= 1
    if i == -1: return arr[::-1]
    j = len(arr) - 1
    while arr[j] <= arr[i]: j -= 1
    arr[i], arr[j] = arr[j], arr[i]
    arr[i+1:] = reversed(arr[i+1:])
    return arr
```

```python
#W9A7 
def canPlaceFlowers(flowerbed, k):
    n = len(flowerbed)
    for i in range(n):
        if flowerbed[i] == 0:
            left = (i == 0) or (flowerbed[i-1] == 0)
            right = (i == n-1) or (flowerbed[i+1] == 0)
            if left and right:
                flowerbed[i] = 1
                k -= 1
                if k == 0:
                    return True
    return k <= 0


```

```python
#W9A8
def is_path_crossing(moves: str) -> bool:
    x, y = 0, 0
    visited = {(0, 0)}
    for move in moves:
        if move == 'R':
            x += 1
        elif move == 'L':
            x -= 1
        elif move == 'U':
            y += 1
        elif move == 'D':
            y -= 1
        if (x, y) in visited:
            return True
        visited.add((x, y))
    return False

 
```

```python
#W9A9 
def Median(nums):
    a, b, c, d, e = nums
    if a > b: a, b = b, a
    if c > d: c, d = d, c
    if a > c: a, c = c, a
    if b > d: b, d = d, b
    if b > c: b, c = c, b
    if c > e: c, e = e, c
    return c
```

```python
#W9A10
import bisect

def findRadius(houses, heaters):
    houses.sort()
    heaters.sort()
    res = 0
    for h in houses:
        
        idx = bisect.bisect_left(heaters, h)
        
        left = h - heaters[idx-1] if idx > 0 else float('inf')
        
        right = heaters[idx] - h if idx < len(heaters) else float('inf')

        nearest = min(left, right)
        res = max(res, nearest)
    return res


```
