```python
#W7A16
from collections import Counter

def mang_dep(a):
    tso = Counter(a)
    xoa = 0
    for x, tso_x in tso.items():
        if x == tso_x:
            continue
        elif tso_x > x:
            xoa += tso_x - x
        else:
            xoa += tso_x

    return "YES" if xoa == 0 else f"NO {xoa}"


a = list(map(int, input().split()))
print(mang_dep(a))
```

```python
#W7A17
def phan_tu(a):

    ptu = None
    dem = 0

    for x in a:
        if dem == 0:
            ptu = x
            dem = 1
        elif x == ptu:
            dem += 1
        else:
            dem -= 1
        
    if ptu is not None and a.count(ptu) > len(a) // 2:
        return ptu
    return None

a = list(map(int, input().split()))
print(phan_tu(a))


```

```python
#W7A19
def merge_sorted_arrays(A, B):
    result = []
    i, j = 0, 0
    
    
    ascending = A[0] <= A[-1]
    
    while i < len(A) and j < len(B):
        if ascending:
            if A[i] <= B[j]:
                result.append(A[i])
                i += 1
            else:
                result.append(B[j])
                j += 1
        else:  
            if A[i] >= B[j]:
                result.append(A[i])
                i += 1
            else:
                result.append(B[j])
                j += 1
    
   
    result.extend(A[i:])
    result.extend(B[j:])
    
    return result


```

```python
#W7A22
from collections import Counter

def most_common_name(names):

    last_names = [name.split()[-1] for name in names]
    freq = Counter(last_names)
    max_count = max(freq.values())
    candidates = [name for name, count in freq.items() if count == max_count]    
    
    return min(candidates)

N=int(input())
names = [input().strip() for _ in range(N)]

print(most_common_name(names))
```

```python
#W7A23
def count_names(s):
    words = s.split()
    return sum(1 for w in words if len(w) >= 2 and w.isalpha() and w.istitle())
```

```python
#W7A24
def remove_adjacent_duplicates(s):
    words = s.split()
    res = [words[0]] if words else []
    for w in words[1:]:
        if w != res[-1]:
            res.append(w)
    return ' '.join(res)
```

```python
#W7A25
def max_k_digits(s, k):
    stack = []
    drop = len(s) - k
    for ch in s:
        while drop and stack and stack[-1] < ch:
            stack.pop()
            drop -= 1
        stack.append(ch)
    return ''.join(stack[:k])
```

```python
#W7A26
import re

def sum_and_avg(s):
    nums = list(map(int, re.findall(r'\d+', s)))
    total = sum(nums)
    avg = total / len(nums) if nums else 0
    return total, round(avg, 2)
```

```python
#W7A27
def ok(s: str) -> bool:
    ngoac_mo = []
    axa = {')': '(', '}': '{', ']': '['}

    for ngoac in s:
        if ngoac in axa:
            ngoac_mo_tu = ngoac_mo.pop() if ngoac_mo else '#'
            if axa[ngoac] != ngoac_mo_tu:
                return False
        else:
            ngoac_mo.append(ngoac)

    return not ngoac_mo



if __name__ == "__main__":
    test_strings = ["()", "()[]{}", "(]", "([)]", "{[]}", "((()))", "({[()]})", "((())", "())"]
    for s in test_strings:
        print(f"{s}: {ok(s)}")

```

```python
#W7A28
def longest_composite_segment(arr, N):
    res, temp = [], []
    for x in arr:
        if x > N:
            temp.append(x)
        else:
            if len(temp) > len(res):
                res = temp[:]
            temp = []
    if len(temp) > len(res):
        res = temp
    return res
```

```python
#W7A29
def longest_increasing_continuous(arr):
    res, temp = [], []
    for i in range(len(arr)):
        if i == 0 or arr[i] > arr[i-1]:
            temp.append(arr[i])
        else:
            if len(temp) > len(res):
                res = temp[:]
            temp = [arr[i]]
    if len(temp) > len(res):
        res = temp
    return res
```

```python
#W7A30
import bisect

def length_of_LIS(arr):
    tails = []
    for x in arr:
        idx = bisect.bisect_left(tails, x)
        if idx == len(tails):
            tails.append(x)
        else:
            tails[idx] = x
    return len(tails)
```
